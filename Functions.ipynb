```python
def add_numbers(a, b):
    return a + b

result = add_numbers(3, 5)

print("Result of addition:", result)
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n + factorial(n-1)

result = factorial(5)
print("factorial of 5:", result)

#This is a self-returning function that uses recursion to achieve a result until reaching a base case where the argument n = 0
#assign the return to a variable and let the repetition process the operation until the process is finished
#Python returns an object reference, so result holds that reference to a known memory address
```

```python
def fibonacci(n):
    fib_series = [0, 1]
    for i in range(2, n+1):
        fib_series.append(fib_series[-1] + fib_series[-2])
    return fib_series
result = fibonacci(3)
print("Fibonacci Series:", result)
```
