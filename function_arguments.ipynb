```python
#Variable length arguments can go beyond just pass by name or pass by value and reference

def print_scores(score, *scores):
    print('Test scores are:')
    print(score)
    for i in scores:
        print(i)
print_scores(10)
print_scores(30,50,53,92)
    
```

```python
def calculate_tax(amount, tax_rate):
    tax = amount * tax_rate
    return tax

def main():
    tax = calculate_tax(85.0, .05) # tax is a local variable
    print('Tax:', tax)

if __name__ == '__main__':
    main()
```
