```python
#Global variable
tax = 10


def calculate_tax(amount, tax_rate):
    global tax
    tax = amount * tax_rate

def main():
    tax = calculate_tax(85.0, .05) # tax is a a local variable
    print('Tax:', tax)

if __name__ == "__main__":
    main()
```

```python
def unit_converter():
    user_input = "y"

    while user_input.lower() == "y":
        feet = float(input("How many feet? "))
        inches = float(input("How many inches? "))

        inches_to_feet = inches / 12
        feet_to_inches = feet * 12

        print(f"{feet} feet is equal to {feet_to_inches} inches")
        print(f"{inches} inches is equal to {inches_to_feet} feet")

        user_input = input("Would you like to continue? (Y/y to continue): ")

    print("Thank you for using the Unit Converter!")


        
unit_converter()

        
    
```
