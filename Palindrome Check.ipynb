```python
# Charlie Assign6
#I wanted to use a basic while loop and not do because there's not background processes
#or recursion happening until base case, I use the standard "enter y" loop to keep the program running


def clean_string(s):
    result = ""
    for ch in s:
        if ch.isalpha():      # keep only letters
            result += ch.lower()
    return result

def is_palindrome(s):
    return s == s[::-1]

def main():
    user_choice = "y"

    while user_choice == "y":
        user_input = input("Enter a string: ")

        cleaned = clean_string(user_input)

        if is_palindrome(cleaned):
            print("The string is a palindrome.")
        else:
            print("The string is not a palindrome.")

        user_choice = input("Do you want to check another string? (y/n): ")
        user_choice = user_choice.strip().lower()

    print("Bye!")

if __name__ == "__main__":
    main()
    
```
