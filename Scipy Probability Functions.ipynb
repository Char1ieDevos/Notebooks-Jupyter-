```python
import scipy.stats

u = scipy.stats.uniform(90, 10)

u.rvs(5)

u.rvs(1000).mean()

(u.rvs(1000) < 92).sum() / 1000.0
```

```python
#Bernoulli trials

#A trial is an experiment which can be repeated many times with the same probabilities, 
#each realization being independent of the others

#success and failure are mutually exclusive

#Example: arrival of requests in a web server per time slot

#geometric distribution or k failed trials until success follows that 
#each probability 1 - p to failure and the num 
# of failures k - 1 must be (1-p)^k-1

#Binomial distribution assumes n num of trials, p successes, and k failures. 
# Pr(X = k) = (n | k)(1-p)^n-k_p^k

```

```python
import scipy.stats

test = scipy.stats.binom(n=100, p=0.1)
test.cdf(1)

test.cdf(10)

test.ppf(0.05)
```

```python
N = 20000
wells = scipy.stats.binom(n=9, p=0.1)
trials = wells.rvs(N)
(trials == 1).sum() / float(N)

#rvs() is a real Monte Carlo simulator for binomial calculations across thousands of trials at runtime
```
